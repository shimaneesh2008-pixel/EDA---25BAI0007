```python
import pandas as pd
```

```python
df1NT = pd.DataFrame({ 'Cinema': ['Pushpa', 'Bahubali', 'Saaho', 'Salaar', 'Kalki', 'Rebel', 'IT', 'Obsession', 'HIT', 'U-TURN', 'ABCD'],
                       'Collection Net(in crores)' : [220, 10000, 8554, 5155, 7981, 9100, 560, 132, 852, 573, 992]})
df2NT =  pd.DataFrame({'Cinema': ['10 Raw (2016)', '9 Go (1999)', '8 Up (2009)', '7 It (2017)', '6 M (1931)', '5 May (2002)' ,'4 Ben (1972)', '3 Pi (1998)'],
                       'Collection Net(in crores)': [98, 93, 44, 77, 69, 56, 31, 53]})

df1GR =  pd.DataFrame({ 'Cinema': ['Pushpa', 'Bahubali', 'Saaho', 'Salaar', 'Kalki', 'Rebel', 'IT', 'Obsession', 'HIT', 'U-TURN', 'ABCD'],
                       'Collection Gross(in crores)' : [320, 10100, 8654, 5255, 8081, 9200, 660, 232, 952, 673, 1092]})
df2GR =  pd.DataFrame({'Cinema': ['10 Raw (2016)', '9 Go (1999)', '8 Up (2009)', '7 It (2017)', '6 M (1931)', '5 May (2002)' ,'4 Ben (1972)', '3 Pi (1998)'],
                       'Collection Gross(in crores)': [93, 44, 78, 97, 87, 89, 39, 43]})
```

```python
dfNT = pd.concat([df1NT, df2NT], ignore_index=True)
dfGR = pd.concat([df1GR, df2GR], ignore_index=True)
```

```python
df = pd.concat([dfGR, dfNT], axis=1)
df
```

```python
dfNT = pd.concat([df1NT, df2NT], ignore_index=True)
dfGR = pd.concat([df1GR, df2GR], ignore_index=True)

df = dfNT.merge(dfGR, how='right')
df
```

```python
dfNT = pd.concat([df1NT, df2NT], ignore_index=True)
dfGR = pd.concat([df1GR, df2GR], ignore_index=True)

df = dfNT.merge(dfGR, how='inner')
df
```

```python
dfNT = pd.concat([df1NT, df2NT], ignore_index=True)
dfGR = pd.concat([df1GR, df2GR], ignore_index=True)

df = dfNT.merge(dfGR, how='left')
df
```
