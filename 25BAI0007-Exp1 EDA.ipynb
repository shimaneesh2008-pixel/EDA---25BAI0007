```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
profit_by_item = df.groupby('Item Type')['Total Profit'].sum().reset_index()
profit_by_item = profit_by_item.sort_values(by='Total Profit', ascending=False)

plt.figure(figsize=(12, 6))
sns.barplot(data=profit_by_item, x='Item Type', y='Total Profit', palette='viridis')
plt.title('Total Profit by Product Category')
plt.xticks(rotation=45)
plt.ylabel('Total Profit ($)')
plt.tight_layout()
plt.show()
```

```python
channel_sales = df.groupby('Sales Channel')['Units Sold'].sum()

plt.figure(figsize=(6, 6))
channel_sales.plot(kind='pie', autopct='%1.1f%%', colors=['#4CAF50', '#FF9800'], startangle=90)
plt.title('Sales Volume Breakdown: Online vs Offline')
plt.ylabel('')
plt.show()
```
